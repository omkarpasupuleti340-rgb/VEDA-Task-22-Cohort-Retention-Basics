```python
import pandas as pd

cohort_df = pd.DataFrame({
    "signup_month": ["2026-01", "2026-01", "2026-02", "2026-02", "2026-03"],
    "order_month": ["2026-01", "2026-02", "2026-02", "2026-03", "2026-03"],
    "active_customers": [2, 1, 2, 1, 1]
})

cohort_df
```

```python
cohort_matrix = cohort_df.pivot(
    index="signup_month",
    columns="order_month",
    values="active_customers"
)

cohort_matrix
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

sns.heatmap(
    cohort_matrix,
    annot=True,
    fmt=".0f",
    cmap="Blues"
)

plt.title("Cohort Retention Heatmap")
plt.xlabel("Order Month")
plt.ylabel("Signup Month")

plt.show()
```

```python
print("Insights:")
print("1. Cohort retention is highest during the initial purchase period.")
print("2. Customer activity changes across subsequent order months.")
print("3. The cohort heatmap helps compare customer retention across signup months.")
```
